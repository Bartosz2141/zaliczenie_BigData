```python
import numpy as np
```

```python
X = np.arange(80).reshape((20, 4))
y = np.random.randint(low=0, high=2, size=20)
```

```python
X.shape
```

```python
y.shape
```

```python
y
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, stratify=y)
```

```python
y_train
```

```python
y_test
```

```python
from sklearn.model_selection import StratifiedShuffleSplit
```

```python
splitter = StratifiedShuffleSplit(n_splits=2, test_size=.2)
```

```python
gen = splitter.split(X, y)
```

```python
for train_idx, test_idx in gen:
  print(X[train_idx])
  print(X[test_idx])
```
